#### CANNY EDGE DETECTION
- `edges=cv2.Canny(image,threshold1,threshold2)`
- THE IMAGE PASSED MUST HAVE TO BE THE GRAY SCALE
- DETECT BORDERS
- SEPERATE OBJECTS
- FEATURE EXTRACTION
- SKETCH EFFECT
- `threshold1` = to detect weak edges
- `threshold2` = to detect strong edges

````python
gray_image=cv.cvtColor(image,cv.COLOR_BGR2GRAY)
canny_edge_gray_image=cv.Canny(gray_image,5,90)
canny_edge_image=cv.Canny(image,5,90)
cv.imshow("CANNY EDGE IMAGE",canny_edge_image)
cv.imshow("CANNY EDGE GRAY IMAGE",canny_edge_gray_image)
cv.waitKey(0)
cv.destroyAllWindows()
````

#### THRESHOLDING

##### BINARY THRESHOLDING
- `ret,thres_img=cv.threshold(image,lower_threshold,maximum_value,method)`

````python
ret,thresh_image=cv.threshold(gray_image,80,220,cv.THRESH_BINARY)
cv.imshow("THRESHOLDED IMAGE",thresh_image)
cv.waitKey(0)
cv.destroyAllWindows()
````

##### ADAPTIVE THRESHOLDING
- It produc the sketchy image
- `binary=cv2.adaptiveThreshold(gray_image,max_value,adaptiveMethod,
threshold_type,block_size,C)`
- the `block_size` determines the size of the local neighborhood used to calculate the threshold
- the `block_size` mush be an odd number greater than 1
- `C` is the constant that is subtracted from the calculated local/weighted mean.

````python
adaptive_img=cv.adaptiveThreshold(
    src=gray_image,
    maxValue=220,
    adaptiveMethod=cv.ADAPTIVE_THRESH_MEAN_C,
    thresholdType=cv.THRESH_BINARY,
    blockSize=11,
    C=3
)
cv.imshow("ADAPTIVE IMAGE",adaptive_img)
cv.waitKey(0)
cv.destroyAllWindows()
````

#### BITWISE OPERATIONS
- COMBINE 2 IMAGES
- CUTOUT THE PART OF ONE IMAGE USING ANOTHER IMAGES
- FLIP SPECIFIC AREA OR REMOVE THEM
- `cv2.bitwise_and(img1,img2)`
- `cv2.bitwise_or(img1,img2)`
- `cv2.bitwise_not(img1)`

````python
import numpy as np
img1=np.zeros((400,400),dtype="uint8")
img2=np.zeros((400,400),dtype="uint8")
cv.circle(img1,(150,150),100,255,-1)
cv.rectangle(img2,(100,100),(300,300),255,-1)
bitwise_and=cv.bitwise_and(img1,img2)
bitwise_or=cv.bitwise_or(img1,img2)
bitwise_not=cv.bitwise_not(img1)
cv.imshow("CIRCLE",img1)
cv.imshow("RECTANGLE",img2)
cv.imshow("AND",bitwise_and)
cv.imshow("OR",bitwise_or)
cv.imshow("NOT",bitwise_not)
cv.waitKey(0)
cv.destroyAllWindows()
````
