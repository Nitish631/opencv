#### GAUSSIAN BLUR
- `blurred_image=cv2.GaussianBlur(image,(kernel_size_x,kernel_size_y),sigma)`
- The `kernel_size_x` and `kernel_size_y` must be odd
- The `sigma` defines how mush the filter effect the image. `0` means it will be random
- USED TO REMOVE THE NOISES
- USED TO SMOOTH THE IMAGE

- THE CONCEPT IS TO ADD ALL THE ELEMENT OF THE KERNEL MATRIX AND THEN FIND THE AVERAGE OF ALL THE PIXEL AND THEN REPLACE THE CENTER PIXEL WITH THE AVERAGE ONE

````python
gaussian_image=cv.GaussianBlur(image,(7,7),0)
cv.imshow("GAUSSIAN BLUR",gaussian_image)
cv.waitKey(0)
cv.destroyAllWindows()
````

#### MEDIAN BLUR
- `blurred=cv2.medianBlur(image,kernel_size)`
- THE CONCEPT IS TO CALCULATE THE MEDIAN VALUE OF THE KERNEL'S ELEMENT AND THEN FILL THAT ELEMENT TO THE MIDDLE POINT

````python
medain_blur=cv.medianBlur(image,11)
cv.imshow("ORIGINAL IMAGE",image)
cv.imshow("MEDIAN BLUR",medain_blur)
cv.waitKey(0)
cv.destroyAllWindows()
````

#### SHARPENING FILTER
- `cv2.filter2D(src,depth,kernel)`

````python
import numpy as np
sharpen_kernel=np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])
shapened_image=cv.filter2D(image,-5,sharpen_kernel)
cv.imshow("ORIGINAL IMAGE",image)
cv.imshow("SHARPENED IMAGE",shapened_image)
cv.waitKey(0)
cv.destroyAllWindows()
````
