# WORKING WITH VIDEO AND WEBCAM

#### INITIALIZATION

```python
import cv2 as cv
image=cv.imread("assets/p3.jpg")
print(image)
```

#### VIDEO CAPTURING
- `cv2.VideoCapture(source)`
- source = 0 for self computer camera
- source = 1 for external web camera

```python
cap=cv.VideoCapture(0)
while True:
    ret,frame=cap.read() # ret=True/False , frame = image
    if not ret:
        print("could not read frame")
        break
    cv.imshow("Webcam Feed",frame)
    if cv.waitKey(1)& 0xFF == ord('q'): ## here 1 means wait for 1 ms and check if q is pressed
        print("Quitting........")
        break
cap.release() ## Close the camera
cv.destroyAllWindows()
```

#### SAVING WEBCAM VIDEO TO A FILE WITH OPENCV
- Steps
  - open camera
  - define the width and height
  - start recording and define the source
  - show the capturing
  - quit if pressed `q`
- `cv2.VideoWriter(filename,codec,fps,frame_size)`
- codec means video formatt . i.e, avi,mp4,xvid
- fps = frame per second
- frame_size = means the size of the frame

```python
camera=cv.VideoCapture(0)
frame_width=int(camera.get(cv.CAP_PROP_FRAME_WIDTH))
frame_height=int(camera.get(cv.CAP_PROP_FRAME_HEIGHT))
codec=cv.VideoWriter_fourcc(*'XVID')
recorder=cv.VideoWriter("assetsO/firstVideo.avi",codec,fps=30,frameSize=(frame_width,frame_height))
while True:
    success,image=camera.read()
    if not success:
        print("Interrupted while capturing the video.")
        break
    recorder.write(image)
    cv.imshow("Recording live",image)
    if cv.waitKey(1)& 0xFF == ord('q'):
        print("Quitting........")
        break
camera.release()
recorder.release()
cv.destroyAllWindows()
```
