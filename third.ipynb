# BASIC IMAGE DRAWING TECHNIQUES

```python
import cv2 as cv
image=cv.imread("assets/p3.jpg",1)
cv.imshow("Image",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

#### LINE DRAWING
- `cv2.line(image,pt1,pt2,color,thickness)`

```python
pt1=(370,350)
pt2=(560,570)
color=(255,0,0)
cv.line(image,pt1,pt2,color,thickness=4)
cv.imshow("LINED IMAGE",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

#### RECTANGLE DRAWING
- `cv2.rectangle(image,pt1,pt2,color,thickness)`

```python
cv.rectangle(image,pt1,pt2,color=(0,255,0),thickness=3)
cv.imshow("RECTANGLED IMAGE",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

#### CRICLE DRAWING
- `cv2.circle(image,center,radius,color,thickness)`
- use the `thickness = -1` to fill up the circle.

```python
center=(730,550)
cv.circle(image,center,radius=110,color=(0,0,255),thickness=4)
cv.imshow("RECTANGLED IMAGE",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

#### ADDING TEXT
- use `cv2.putText()`
- `cv2.putText(image,text,org,font,fonntScale,color,tickness)`
- `org` = BOTTOM LEFT POINT OF THE TEXT.

```python
cv.putText(
    img=image,
    text="Hello python programmers",
    org=(450,250),
    fontFace=cv.FONT_HERSHEY_TRIPLEX,
    fontScale=1.5,
    color=(255,255,3))
cv.imshow("TEXT ADDED IMAGE",image)
cv.waitKey(0)
cv.destroyAllWindows()
```
