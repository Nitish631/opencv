## INITIALIZATION

```python
import cv2 as cv
import matplotlib.pyplot as plt
```

## LOAD IMAGES

```python
image = cv.imread("assets/p1.jpg",1)
image
```

## DISPLAY IMAGES

```python
cv.imshow("NITISH IMAGE TITLE",image)
cv.waitKey(0)## WAIT UNTIL ANY KEY IS PRESSED
cv.destroyAllWindows()
```

## SAVE IMAGES

```python
if image is not None:
    success=cv.imwrite("assetsO/po1.jpg",image)
    if success:
        print("IMAGE SAVED SCCESSFULLY.")
    else:
        print("FAILED TO SAVE THE IMAGE")
else:
    print("ERROR: COULD NOT LOAD IMAGE")
```
