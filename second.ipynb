# IMAGE TRANSFORMATION AND MANIPULATION

## INITIALIZATION

```python
import cv2 as cv
```

```python
image=cv.imread("assets/p2.jpg",1)
cv.imshow("image title",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

## RESIZE FUNCITON `cv2.resize()`
- `cv2.resize(immage,(width,height),fx,fy,interpolation)`

```python
resized=cv.resize(image,(300,300))
cv.imshow("Resized title",resized)
cv.waitKey(0)
cv.destroyAllWindows()
cv.imwrite("assetsO/po2.jpg",resized)
```

```python
resized=cv.resize(image,None,fx=1,fy=0.5)
cv.imshow("Resized title",resized)
cv.waitKey(0)
cv.destroyAllWindows()
```

- FOR THE INTERPOLATION,
  - `cv.INTER_NEATEST`
  - `cv.INTER_LINEAR`
  - `cv.INTER_CUBIC`
  - `cv.INTER_AREA`
  - `cv.INTER_LANCZOS4` high quality interpolation using and 8x8 neighborhood.

```python
resized=cv.resize(image,(500,500),interpolation=cv.INTER_LINEAR)
cv.imshow("Resized title",resized)
cv.waitKey(0)
cv.destroyAllWindows()
```

## CROPPING IMAGES USING SLICING IN OPEN CV

```python
cropped_image=image[100:300,300:500]## height , width
cv.imshow("Cropped image title",cropped_image)
cv.waitKey(0)
cv.destroyAllWindows()
```

- resizing the cropped image

```python
resized_cropped=cv.resize(cropped_image,(600,600))
cv.imshow("Resized cropped image title",resized_cropped)
cv.waitKey(0)
cv.destroyAllWindows()
```

## IMAGE ROTATION AND FLIPPING
#### ROTATION
- `M=cv2.getRotationMatrix2D(center,angle,scale)`
- `rotated_image=cv2.wrapAffine(image,M,(width,height))`


```python
height,width,channel=image.shape
M=cv.getRotationMatrix2D(center=(height//2,width//2),angle=-45,scale=2)
print(M)
rotated_image=cv.warpAffine(image,M,(500,500))
if rotated_image is not None:
    cv.imshow("Rotated image",rotated_image)
    cv.waitKey(0)
    cv.destroyAllWindows()
```

#### FLIPPING
- `cv2.flip(image,flipcode)`
    - flipcode = 0 `VERTICAL`
    - flipcode = 1 `HORIZONTAL`
    - flipcode = -1 `BOTH HORIZONTAL AND VERTICAL`

```python
fliped_image0=cv.flip(image,0)
fliped_image1=cv.flip(image,1)
fliped_image2=cv.flip(image,-1)
cv.imshow("FLIPED IMAGE 0",fliped_image0)
cv.imshow("FLIPED IMAGE 1",fliped_image1)
cv.imshow("FLIPED IMAGE 2",fliped_image2)
cv.waitKey(0)
cv.destroyAllWindows()
```
